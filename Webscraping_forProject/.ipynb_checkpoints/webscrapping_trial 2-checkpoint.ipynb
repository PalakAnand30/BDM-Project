```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import random
import time

reviewlist = []

def extractReviews(reviewUrl, pageNumber):
    headers = {'User-Agent':"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"}
    resp = requests.get(reviewUrl, headers=headers)
    
    # Debugging: Ensure the page is being fetched correctly
    if resp.status_code != 200:
        print(f"Failed to retrieve page {pageNumber}. Status code: {resp.status_code}")
        return
    
    soup = BeautifulSoup(resp.text, 'html.parser')
    
    # Debugging: Check the soup content to verify it's correct
    print(f"Scraping page {pageNumber}. Length of page content: {len(soup)}")
    
    reviews = soup.findAll('div', {'data-hook': "review"})
    
    # Debugging: Check if reviews are found
    if not reviews:
        print(f"No reviews found on page {pageNumber}")
        return
    
    for item in reviews:
        try:
            review_title = item.find('a', {'data-hook':"review-title"})
            review_body = item.find('span', {'data-hook': 'review-body'})
            review_rating = item.find('i', {'data-hook': 'review-star-rating'})
            
            # Debugging: Print to check if each element is found
            print(f"Review Title: {review_title}")
            print(f"Review Body: {review_body}")
            print(f"Review Rating: {review_rating}")

            review = {
                'productTitle': soup.title.text.replace("Amazon.in:Customer reviews: ", "").strip(),
                'Review Title': review_title.span.text.strip() if review_title else "N/A",
                'Rating': review_rating.text.strip() if review_rating else "N/A",
                'Review Body': review_body.text.strip() if review_body else "N/A"
            }
            reviewlist.append(review)
        except AttributeError as e:
            # Handle missing attributes gracefully
            print(f"Error processing review on page {pageNumber}: {e}")
            continue

def totalPages(productUrl):
    headers = {'User-Agent':"Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"}
    resp = requests.get(productUrl, headers=headers)
    soup = BeautifulSoup(resp.text, 'html.parser')
    
    reviews = soup.find('div', {'data-hook':"cr-filter-info-review-rating-count"})
    
    if reviews is None:
        print("Couldn't find the total review count. Returning a default number of pages (100).")
        return 100
    
    str_num_review = reviews.text.strip().split(', ')[1].split(" ")[0]
    num_review = str_num_review.replace(',', '')
    num_review = int(num_review)
    
    return num_review // 10  # Assuming 10 reviews per page

def main():
    productUrl = "https://www.amazon.in/Omron-Automatic-Intellisense-Technology-Measurement/product-reviews/B00F38B3NW/"
    
    totalPg = totalPages(productUrl)
    print(f"Total pages: {totalPg}")
    
    for i in range(1, totalPg + 1):
        print(f"Scraping page {i}")
        reviewUrl = productUrl + f"?pageNumber={i}"
        
        try:
            extractReviews(reviewUrl, i)
            time.sleep(random.uniform(1, 3))  # Random sleep to avoid blocking
        except Exception as e:
            print(f"Error on page {i}: {e}")
            break

    # Debugging: Check if any reviews were scraped
    print(f"Total reviews scraped: {len(reviewlist)}")
    
    if reviewlist:
        df = pd.DataFrame(reviewlist)
        df.to_csv('output.csv', index=False)
        print("Reviews successfully saved to output.csv")
    else:
        print("No reviews scraped. CSV file not created.")

main()
```
